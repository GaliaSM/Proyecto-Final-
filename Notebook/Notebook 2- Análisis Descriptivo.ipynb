## 1.Importación de las librerías y del daset limpio

- Se importan las librerías que se van a necesitar y los datos limpios.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../Datos_Limpios/Proyecto_Final_Datos_Limpios.csv")
```

## 2. Información general del dataset

- Cuando se exporta el dataet CSV hay que tener en cuenta que las columnas que antes se pasaron a categoria se vuelven a transformar a texto (str).
- Se puede ver info() 

```python

df.info()
```

## 3. Reconvertir las columnas al tipo category

```python
cols_to_category = ['hotel',"meal","country_ISO-3","market_segment",
                    "distribution_channel","reserved_room_type",
                    "assigned_room_type","deposit_type","customer_type",
                    "reservation_status","Country Name","continent",
                    "family","room_change"]

for col in cols_to_category:
    if col in df.columns:
        df[col] = df[col].astype('category')
```

## 7. Histogramas de variables numéricas

- Se aplican  los histogramas porque esto nos permite ver la distribución de las variables numéricas.
- Se puede detectar si las variables tienen valores extremos o sesgos.
- Se crea una lista con las columnas que se quieren analizar.
- Se aplica "hist(figsize=(8,4), bins=30)" para obtener el histograma

```python
num_cols = ['lead_time','total_stay','total_people','adr']

df[num_cols].hist(figsize=(8,4), bins=30)
plt.suptitle("Distribución de variables numéricas")
plt.show()
```

## 8. Graficos simples 

- Se aplican graficos de barras porque estos ayudan a entender la frecuencia de las categorias más importantes

### 1. Cancelaciones

- Se analizan cuantas reservas se cancelan y cuantas no  (is_canceled = 1) y no canceladas (is_canceled = 0).

- Codigo:

- .figure indicar el tamaño del gráfico 
- .countplot es una funcion de la libreria de seaborn se usa para crear gráficos de barras
- .title indicar le titulo del gráfico.
- .show mostrar

```python
plt.figure(figsize=(8,4))
sns.countplot(data=df, x='is_canceled')
plt.title("Cancelaciones")
plt.show()


```

### 2. Reservas por tipo de hotel 

- Se compara la cantidad de reservas entre City Hotel y Resort Hotel.

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='hotel')
plt.title("Reservas por tipo de hotel")
plt.show()
```

### 3. Reservas por continente

-  Se  analiza la distribcion las reservas según el continente de origen del cliente

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='continent')
plt.title("Reservas por continente")
plt.xticks(rotation=45)
plt.show()

```

### 4. Reservas familia y no familia

- Se compara la cantidad de reservas realizadas por familias (si) y no familias (no).

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='family')
plt.title("Reservas: familias vs no familias")
plt.show()
```

### 5. Reservas Mensuales 

- Se analiza como se comportan las resrvas duarante los meses ( Enero =1, Febreo= 2...) se analiza si hay un patron de comportamiento, estacionalidad.

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='arrival_month_num')
plt.title("Reservas Mensuales")
plt.show()
```
